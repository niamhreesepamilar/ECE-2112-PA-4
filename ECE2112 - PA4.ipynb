```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
board = pd.read_excel('board2.xlsx')
board
```

```python
df = board.copy()

df['Average'] = df[['Math', 'Electronics', 'GEAS', 'Communication']].mean(axis=1)
df
```

```python
### A. VISAYAS COMMUNICATION DATAFRAME
```

```python
VisComm = df.loc[(df['Hometown'] == 'Visayas') & (df['Track'] == 'Communication'),
    ['Name', 'Gender', 'Math', 'Electronics', 'Average']]

VisComm
```

```python
len(VisComm)
```

```python
### B. VISAYAS FEMALE DATAFRAME
```

```python
VisFemale = df.loc[(df['Hometown'] == 'Visayas') & (df['Gender'] == 'Female'),
    ['Name', 'Track', 'GEAS', 'Electronics', 'Average']]
VisFemale
```

```python
VisFemale.loc[VisFemale['Average'] >= 60]
```

```python
### C. CATEGORY-AVERAGE VISUALIZATION
```

```python
track_mean = df.groupby('Track')['Average'].mean()
track_mean
```

```python
gender_mean = df.groupby('Gender')['Average'].mean()
gender_mean
```

```python
hometown_mean = df.groupby('Hometown')['Average'].mean()
hometown_mean
```

```python
plt.figure(figsize=(15, 4))

plt.subplot(1, 3, 1)
plt.bar(track_mean.index, track_mean.values)
plt.title('Mean Average by Track')
plt.xlabel('Track')
plt.ylabel('Mean Average')
plt.xticks(rotation=20)

plt.subplot(1, 3, 2)
plt.bar(gender_mean.index, gender_mean.values)
plt.title('Mean Average by Gender')
plt.xlabel('Gender')
plt.ylabel('Mean Average')

plt.subplot(1, 3, 3)
plt.bar(hometown_mean.index, hometown_mean.values)
plt.title('Mean Average by Hometown')
plt.xlabel('Hometown')
plt.ylabel('Mean Average')

plt.tight_layout()
plt.show()
```
